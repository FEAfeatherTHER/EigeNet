```python
os.chdir(f"/data/250010171/code/EigeNet_discriminant/")
from models.loss.evaluator import Evaluator
metric_calculator = Evaluator()
from models.layers.utils import compute_metrics
gt_rir_path = '/data/250010171/code/EigeNet_discriminant/visualization/out/gt_ir.wav'
pred_rir_path = '/data/250010171/code/EigeNet_discriminant/visualization/out/pred_ir_ours_wo_align.wav'
gt_rir = librosa.load(gt_rir_path, sr=16000)[0]
pred_rir = librosa.load(pred_rir_path, sr=16000)[0]
min_length = min(len(gt_rir), len(pred_rir))
gt_rir = torch.from_numpy(gt_rir[:min_length]).unsqueeze(0).unsqueeze(0).float().numpy()
pred_rir = torch.from_numpy(pred_rir[:min_length]).unsqueeze(0).unsqueeze(0).float().numpy()
edt_error, c50_error, t60_relative_error, _ = compute_metrics(gt_rir, pred_rir, metric_calculator)

edt_error = round(np.mean(edt_error), 3)
c50_error = round(np.mean(c50_error), 3)
t60_relative_error = round(np.mean(t60_relative_error), 3)
print(f"EDT Error: {edt_error}")
print(f"C50 Error: {c50_error}")
print(f"T60 Error: {t60_relative_error}")
```
